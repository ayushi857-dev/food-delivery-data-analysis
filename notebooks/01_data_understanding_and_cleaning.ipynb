```python
# Counting decimal-style time values

decimal_order = df["Time_Orderd"].str.match(
    r"^\d+(\.\d+)?$",
    na=False
)

decimal_pickup = df["Time_Order_picked"].str.match(
    r"^\d+(\.\d+)?$",
    na=False
)

print("Decimal values in order time:", decimal_order.sum())
print("Decimal values in pickup time:", decimal_pickup.sum())
```

```python
# Converting decimal time values into normal time format

def convert_time(value):
    if pd.isna(value):
        return value

    try:
        number = float(value)

        if 0 <= number <= 1:
            total_minutes = round(number * 24 * 60)

            hours = (total_minutes // 60) % 24
            minutes = total_minutes % 60

            return f"{hours:02d}:{minutes:02d}"

    except ValueError:
        pass

    return value


df["Time_Orderd"] = df["Time_Orderd"].apply(convert_time)
df["Time_Order_picked"] = df["Time_Order_picked"].apply(convert_time)

print("Order time examples:")
print(df["Time_Orderd"].head(15))

print("\nPickup time examples:")
print(df["Time_Order_picked"].head(15))
```

```python
# Checking remaining time values

print("Order time values:")
print(df["Time_Orderd"].value_counts().head(15))

print("\nPickup time values:")
print(df["Time_Order_picked"].value_counts().head(15))
```

```python
# Checking valid time formats

order_check = pd.to_datetime(
    df["Time_Orderd"],
    format="%H:%M",
    errors="coerce"
)

pickup_check = pd.to_datetime(
    df["Time_Order_picked"],
    format="%H:%M",
    errors="coerce"
)

print("Order time missing:", df["Time_Orderd"].isna().sum())
print("Order time invalid:", (
    df["Time_Orderd"].notna() & order_check.isna()
).sum())

print("\nPickup time missing:", df["Time_Order_picked"].isna().sum())
print("Pickup time invalid:", (
    df["Time_Order_picked"].notna() & pickup_check.isna()
).sum())
```

```python
# Checking pickup times that could not be converted

invalid_pickup = df.loc[
    df["Time_Order_picked"].notna() & pickup_check.isna(),
    "Time_Order_picked"
]

print("Total unusual pickup records:", len(invalid_pickup))

print("\nDifferent unusual time values:")
print(invalid_pickup.value_counts().head(30))
```

```python
# Correcting pickup times after midnight

df["Time_Order_picked"] = df["Time_Order_picked"].replace({
    "24:05:00": "00:05",
    "24:10:00": "00:10",
    "24:15:00": "00:15"
})

print("Corrected values:")
print(df["Time_Order_picked"].value_counts().head(10))
```

```python
# Final time validation

order_check = pd.to_datetime(
    df["Time_Orderd"], format="%H:%M", errors="coerce"
)

pickup_check = pd.to_datetime(
    df["Time_Order_picked"], format="%H:%M", errors="coerce"
)

print("Order time invalid:", (
    df["Time_Orderd"].notna() & order_check.isna()
).sum())

print("Pickup time invalid:", (
    df["Time_Order_picked"].notna() & pickup_check.isna()
).sum())
```

### Time Data Cleaning

- Converted decimal time values into HH:MM format.
- Found 1,731 missing values in Time_Orderd.
- Normalized 880 unusual pickup-time records:
  24:05:00 → 00:05
  24:10:00 → 00:10
  24:15:00 → 00:15
- Rechecked both time columns after cleaning.

The original raw dataset was preserved.

```python
# Checking missing values after time cleaning

missing_values = df.isnull().sum()

missing_report = pd.DataFrame({
    "Missing Values": missing_values,
    "Percentage": (missing_values / len(df) * 100).round(2)
})

print(missing_report[missing_report["Missing Values"] > 0])
```

```python
# Filling missing numerical values with median

df["Delivery_person_Age"] = df["Delivery_person_Age"].fillna(
    df["Delivery_person_Age"].median()
)

df["Delivery_person_Ratings"] = df["Delivery_person_Ratings"].fillna(
    df["Delivery_person_Ratings"].median()
)

df["multiple_deliveries"] = df["multiple_deliveries"].fillna(
    df["multiple_deliveries"].median()
)


# Filling missing categorical values

df["Weather_conditions"] = df["Weather_conditions"].fillna("Unknown")

df["Road_traffic_density"] = df["Road_traffic_density"].fillna("Unknown")

df["Festival"] = df["Festival"].fillna("Unknown")

df["City"] = df["City"].fillna("Unknown")


# Checking remaining missing values

print(df.isnull().sum())
```
